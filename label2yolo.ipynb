```python
from pathlib import Path
import shutil
import base64
import json
from tqdm.notebook import tqdm

```

```python
names = ["Book", "Inkiness", "Vitium", "Crease", "defaced", "Patch", "Signature"]
id2name = {i: name for i, name in enumerate(names)}
name2id = {name: i for i, name in enumerate(names)}
print(id2name)
print(name2id)
```

```python
JSON_PATH = Path(r"D:\books-dataset\yolo-exp-etl-data\train\json_only")
OUTPUT_PATH = Path(r"D:\books-dataset\yolo-exp-etl-data\train\output-yolo")

if not OUTPUT_PATH.exists():
    OUTPUT_PATH.mkdir()
else:
    raise RuntimeError(f"Output path {OUTPUT_PATH} already exists.")
```

```python
def save_base64_image(base64_str, output_path):
    with open(output_path, "wb") as f:
        f.write(base64.b64decode(base64_str))
```

```python
def save_shapes_label(shapes, image_height, image_width, output_path):
    with open(output_path, "w", encoding="utf-8") as f:
        for shape in shapes:
            coord_list = []
            label_id = name2id[shape['label']]
            coord_list.append(label_id)
            for point in shape['points']:
                coord_list.append(point[0] / image_width)
                coord_list.append(point[1] / image_height)
            
            f.write(" ".join(map(str, coord_list)) + "\n")
```

```python
for json_file in tqdm(JSON_PATH.glob("*.json")):
    with open(json_file, "r", encoding="utf-8") as jsonf:
        json_obj = json.load(jsonf)
    save_base64_image(json_obj['imageData'], OUTPUT_PATH / json_obj['imagePath'])
    save_shapes_label(json_obj['shapes'], json_obj['imageHeight'], json_obj['imageWidth'], OUTPUT_PATH / (json_file.stem + ".txt"))
```
