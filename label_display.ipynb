```python
IMAGE_URL = r"D:\论文数据集\文物标注\yolo-exp-etl-data\yolo-exp-etl-data\train\images\00a267879fe83bd5_b.jpg"
LABEL_URL = r"D:\论文数据集\文物标注\yolo-exp-etl-data\yolo-exp-etl-data\train\images\00a267879fe83bd5_b.json"
```

```python
from PIL import Image, ImageDraw, ImageFont
import json
import numpy as np
import json
```

```python
label_json = json.load(open(LABEL_URL, 'r', encoding='utf-8'))
```

```python
label_json
```

```python
label2 = label_json['shapes'][1]
label2
```

```python
image = Image.open(IMAGE_URL).convert('RGBA')
```

```python
points = [(x,y) for x, y in label2['points']]
```

```python
mask = Image.new("RGBA", image.size, (0, 0, 0, 0))  # 初始全透明
draw = ImageDraw.Draw(mask)
draw.polygon(points, outline='blue', width=4, fill=(255, 0, 0, 128))
result = Image.alpha_composite(image, mask)
result
```
